```python
creer_slide_nav(
    section_id="methode_construction",
    section_num="01",
    section_titre="Introduction",
    sous_titre="Méthode de Construction : L'Itération <br>  <br> (Iterated Functions System IFS)",
    contenu="""

 <p>Parmi les méthodes de construction de fractales, on trouve la méthode appelée 
 <span style="color: green; font-weight: bold;"> "Iterated Functions System" (IFS)</span> :
Il s'agit de partir d'une fonction holomorphe $f$ sur le plan complexe, un polynôme par exemple, 
puis de considérer <span style="color: #e74c3c; font-weight:
bold;">la suite des iterées</span> d'un nombre complexe donné par compositions successives de $f$ 
<span style="color: #e74c3c; font-weight: bold;">(orbite) </span><p>
    
    <p><span style="color: #e74c3c; font-weight: bold;"> On se limite au cas où la fonction $f$
    est le polynôme du second degré:</span>
    <p><span style="font-size: 20px;">➥</span> $f_c(z) := z^2 + c,$ avec $c\in \mathbb{C}$ 
    est un paramètre. </span><p>

    <p>Ceci revient à  considérer la suite de nombres complexes:<p>
       
    <div style="background-color: #f0f7ff; padding: 10px; border: 2px dashed #2e86c1; border-radius: 15px; 
    text-align: center; margin: 12px 0;">
        <span style="font-size: 20px; color: #2c3e50;">
            $$z_0\in \mathbb{C},\;\;z_{n+1} = z_n^2 + c, n\geq 0.$$
        </span>
    </div>

<div style="background-color: #2c3e50; padding: 20px; border: 2px dashed #2e86c1; border-radius:
15px; text-align: center; margin: 12px 0;">
        <span style="font-size: 18px; color: #f0f7ff;">
            Soit $z_0$ un point de $\mathbb{C}$. On définit la suite $(z_n)_{n\geq0}$ par :
$z_{n+1} = f_c(z_n) =f_c^{n+1}(z_0)$.
On appelle cette suite l’orbite de $z_0$ sous l’action de $f_c$ et on la note $\mathcal{O}^+(z_0).$
        </span>
    </div>

    <p>Le processus de construction de nos fractales qu'on notera par $M$ et $J_c$ suit trois étapes fondamentales :</p>
    
    <div style="display: flex; flex-direction: column; gap: 10px;">
        <div style="background-color: #fcfcfc; padding: 10px; border-left: 5px solid #2e86c1;">
            <b>1. Initialisation :</b> On choisit un point de départ $z_0$ dans le plan complexe.
        </div>
        
        <div style="background-color: #fcfcfc; padding: 10px; border-left: 5px solid #2e86c1;">
            <b>2. Itération :</b> On applique la fonction de manière répétée 
            ($z_0$ $&#10142;$ $z_1$ $&#10142;$  $z_2$ $&#10142;$ $z_3$ $&#10142;$ $\dots$).
        </div>
        
        <div style="background-color: #fcfcfc; padding: 10px; border-left: 5px solid #2e86c1;">
            <b>3. Test de divergence :</b> On observe si la suite reste bornée ou s'échappe vers 
            l'infini (vitesse d'échappement).
        </div>
    </div>

    ➥ Conséquence sur la couleur qu'on affectera au point $z_0$:
    
    <div style="margin-top: 15px; font-style: italic; color: #7f8c8d; font-size: 14px;">
        ➥ C'est la vitesse à laquelle les points s'échappent 
        qui détermine la couleur de chaque 
       <p> <span style="color: #e74c3c; font-weight: bold;"> 
       $\quad$ pixel (le point d'affixe $z_0$) </span>
        .<p>
    </div>
    """,
    prev_id="proprietes_fractales"
)
```

```python
# --- On définit le contenu avec le code source formaté ---
contenu_impl = r"""
<p>L'implémentation repose sur une approche <b>vectorisée</b> avec NumPy pour maximiser les performances :</p>

<div style="background-color: #2d2d2d; color: #f8f8f2; padding: 15px; border-radius: 10px; font-family: 'Courier New', Courier, monospace; font-size: 0.85em; overflow-x: auto; line-height: 1.4;">
<span style="color: #66d9ef;">def</span> <span style="color: #a6e22e;">mandelbrot</span>(h, w, max_iter=<span style="color: #ae81ff;">50</span>):<br>
&nbsp;&nbsp;&nbsp;&nbsp;y, x = np.ogrid[-<span style="color: #ae81ff;">1.4</span>:<span style="color: #ae81ff;">1.4</span>:h*<span style="color: #ae81ff;">1j</span>, -<span style="color: #ae81ff;">2</span>:<span style="color: #ae81ff;">0.8</span>:w*<span style="color: #ae81ff;">1j</span>]<br>
&nbsp;&nbsp;&nbsp;&nbsp;c = x + y*<span style="color: #ae81ff;">1j</span><br>
&nbsp;&nbsp;&nbsp;&nbsp;z = c<br>
&nbsp;&nbsp;&nbsp;&nbsp;divertime = np.zeros(z.shape, dtype=int)<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #f92672;">for</span> i <span style="color: #f92672;">in</span> <span style="color: #66d9ef;">range</span>(max_iter):<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;z = z**<span style="color: #ae81ff;">2</span> + c<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;diverge = z*np.conj(z) > <span style="color: #ae81ff;">4</span><br>            
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;escaping_now = diverge & (divertime == <span style="color: #ae81ff;">0</span>)<br>  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;divertime[escaping_now] = i<br>           
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;z[diverge] = <span style="color: #ae81ff;">2</span> <span style="color: #75715e;"># Évite l'overflow</span><br><br>
&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #f92672;">return</span> divertime
</div>

<div style="margin-top: 15px; display: flex; gap: 10px;">
    <div style="flex: 1; border-left: 4px solid #2e86c1; padding-left: 10px; font-size: 0.9em;">
        <b>Vectorisation :</b> Le calcul est effectué sur toute la matrice <code>c</code> simultanément, ce qui est bien plus rapide qu'une boucle sur chaque pixel.
    </div>
</div>
"""

# --- 1. Création de la slide avec le code dans le contenu ---
creer_slide_nav(
    section_id="code_mandelbrot",
    section_num="02",
    section_titre="Analyse & Conception",
    sous_titre="Implémentation de la fonction Mandelbrot",
    contenu=contenu_impl,
    prev_id="algorithme_echappement2"
)

# --- 2. Exécution silencieuse pour afficher le résultat visuel en dessous ---
import numpy as np
import matplotlib.pyplot as plt

def mandelbrot(h, w, max_iter=50):
    y, x = np.ogrid[-1.4:1.4:h*1j, -2:0.8:w*1j]
    c = x + y*1j
    z = c
    divertime = np.zeros(z.shape, dtype=int)
    for i in range(max_iter):
        z = z**2 + c
        diverge = z*np.conj(z) > 4            
        escaping_now = diverge & (divertime == 0)  
        divertime[escaping_now] = i           
        z[diverge] = 2 
    return divertime

plt.figure(figsize=(10, 14))
result = mandelbrot(400, 700)
plt.imshow(result, extent=[-2, 0.8, -1.4, 1.4], cmap='magma')
plt.axis('off')
plt.title("Rendu généré par le code ci-dessus", fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import io
import base64

# --- 1. GÉNÉRATION DU GRAPHIQUE DE LA CARDIOÏDE ---
def get_cardioid_plot():
    t = np.linspace(0, 2*np.pi, 200)
    # Équation paramétrique de la frontière de la cardioïde de Mandelbrot
    # c = (1/2)exp(it) - (1/4)exp(2it)
    c_re = 0.5 * np.cos(t) - 0.25 * np.cos(2*t)
    c_im = 0.5 * np.sin(t) - 0.25 * np.sin(2*t)
    
    plt.figure(figsize=(5, 5), facecolor='none')
    plt.plot(c_re, c_im, color='#e74c3c', linewidth=2, label="Frontière $\rho=1$")
    plt.fill(c_re, c_im, color='#fdf2f2', alpha=0.5)
    plt.axhline(0, color='black', lw=0.5)
    plt.axvline(0, color='black', lw=0.5)
    plt.title("Lieu des points fixes attractifs", fontsize=10)
    plt.axis('equal')
    plt.grid(True, linestyle='--', alpha=0.3)
    
    buf = io.BytesIO()
    plt.savefig(buf, format='png', bbox_inches='tight')
    plt.close()
    return base64.b64encode(buf.getvalue()).decode('utf-8')

card_img = get_cardioid_plot()

# --- 2. AFFICHAGE DE LA SLIDE ---
creer_slide_nav(
    section_id="analyse_cardioide",
    section_num="04",
    section_titre="Analyse Dynamique Profonde",
    sous_titre="La Cardioïde : Condition de Stabilité",
    contenu=f"""
    <div style="display: flex; gap: 20px; align-items: start;">
        <div style="flex: 1.2;">
            <p>Le bulbe principal de Mandelbrot est une <b>cardioïde</b>. Elle représente les paramètres $c$ pour lesquels $f_c(z)$ possède un <b>point fixe attractif</b>.</p>
            
            <div style="background: #f8f9fa; padding: 15px; border-radius: 10px; border-left: 5px solid #34495e;">
                <b style="color: #2c3e50;">Démonstration rapide :</b><br>
                1. Point fixe : $z^2 + c = z \implies z^2 - z + c = 0$<br>
                2. Stabilité : $|f'_c(z)| < 1 \implies |2z| < 1$<br>
                3. À la limite ($|2z|=1$), on pose $z = \\frac{1}{2}e^{{i\\theta}}$
            </div>
            
            <p style="margin-top: 10px;">En injectant $z$ dans l'équation du point fixe, on obtient le lieu géométrique :</p>
            <p style="text-align: center; font-size: 16px; color: #e74c3c;">
                $c = \\frac{1}{2}e^{{i\\theta}} - \\frac{1}{4}e^{{2i\\theta}}$
            </p>
        </div>
        
        <div style="flex: 0.8; text-align: center;">
            <img src="data:image/png;base64,{card_img}" style="width: 100%; border-radius: 10px;">
            <p style="font-size: 11px; color: #7f8c8d; margin-top: 5px;">Frontière théorique de la zone de convergence</p>
        </div>
    </div>
    """,
    prev_id="mecanisme_transformation"
)
```

```python
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio

# --- 1. GÉNÉRATION DE L'ANIMATION ---
def get_plotly_html_string():
    theta = np.linspace(0, 2*np.pi, 100)
    groups = [
        {'label': 'Points', 'color': 'green', 'mode': 'markers',
         'z': [0.8 * np.exp(1j * a) for a in np.linspace(0, 2*np.pi, 8, endpoint=False)]},
        {'label': 'Rayons', 'color': 'red', 'mode': 'lines',
         'z': [np.linspace(0, 1.5, 50) * np.exp(1j * a) for a in np.linspace(0, 2*np.pi, 8, endpoint=False)]},
        {'label': 'Grille HV', 'color': 'cyan', 'mode': 'lines',
         'z': [np.linspace(-1.5, 1.5, 50) + 1j*y for y in [0.5, -0.5]] + [x + 1j*np.linspace(-1.5, 1.5, 50) for x in [0.5, -0.5]]},
        {'label': 'Cercles', 'color': 'blue', 'mode': 'lines',
         'z': [r * np.exp(1j * theta) for r in [0.5, 1.0, 1.5]]}
    ]

    all_frames = []
    for current_idx in range(len(groups)):
        for t in np.linspace(0, 1, 15): # Réduit à 15 pour la fluidité sur navigateur
            frame_data = []
            for i, gp in enumerate(groups):
                t_val = 1 if i < current_idx else (t if i == current_idx else 0)
                color = gp['color'] if i <= current_idx else 'lightgrey'
                for z_val in gp['z']:
                    w = z_val**(1 + t_val)
                    x, y = (w.real.tolist(), w.imag.tolist()) if isinstance(w, np.ndarray) else ([w.real], [w.imag])
                    frame_data.append(go.Scatter(x=x, y=y, mode=gp['mode'], 
                                                 line=dict(color=color, width=3), 
                                                 marker=dict(color=color, size=8), 
                                                 showlegend=False))
            all_frames.append(go.Frame(data=frame_data))

    fig = go.Figure(data=all_frames[0].data, frames=all_frames)
    
    fig.update_layout(
        width=700, height=500,
        xaxis=dict(range=[-3, 3], scaleanchor="y", scaleratio=1),
        yaxis=dict(range=[-3, 3]),
        margin=dict(l=10, r=10, t=10, b=10),
        updatemenus=[{
            "type": "buttons",
            "buttons": [{"label": "▶ Transformer", "method": "animate", "args": [None, {"frame": {"duration": 50}}]}]
        }]
    )
    
    # On exporte en string HTML complet (incluant la bibliothèque Plotly.js)
    # include_plotlyjs='cdn' permet d'alléger la slide
    return pio.to_html(fig, full_html=False, include_plotlyjs='cdn')

# Capture du code HTML
plotly_html_code = get_plotly_html_string()

# --- 2. AFFICHAGE DE LA SLIDE ---
creer_slide_nav(
    section_id="dynamique_plotly_z2_large",
    section_num="04",
    section_titre="Analyse Dynamique Profonde",
    sous_titre="Analyse Visuelle de la Transformation $z \\mapsto z^2$",
    contenu=f"""
    <div style="display: flex; flex-direction: column; align-items: center;">
        <div style="width: 100%; border: 2px solid #34495e; border-radius: 12px; overflow: hidden; background: white; box-shadow: 0 10px 30px rgba(0,0,0,0.15);">
            <iframe srcdoc='{plotly_html_code.replace("'", "&apos;")}' 
                    width="100%" height="550px" style="border: none;"></iframe>
        </div>

        <div style="display: flex; gap: 20px; width: 100%; margin-top: 15px;">
            <div style="flex: 1; background: #e8f5e9; padding: 12px; border-radius: 8px; border-top: 4px solid #2e7d32; font-size: 13px;">
                <b style="color: #2e7d32;">Observation :</b> La transformation préserve l'orthogonalité locale (conformalité) mais courbe l'espace globalement.
            </div>
            <div style="flex: 1; background: #f0f7ff; padding: 12px; border-radius: 8px; border-top: 4px solid #2e86c1; font-size: 13px;">
                <b style="color: #2e86c1;">Application :</b> Ce mécanisme de repliement est responsable de la structure dense des ensembles de Julia.
            </div>
        </div>
    </div>
    """,
    prev_id="analyse_cardioide"
)
```

```python
creer_slide_nav(
    section_id="demonstration_mathematique_complete",
    section_num="04",
    section_titre="Analyse Dynamique Profonde",
    sous_titre="Analyse de l'application $f(z) = z^2$",
    contenu="""
    <div style="display: flex; gap: 20px; align-items: stretch;">
        
        <div style="flex: 1; display: flex; flex-direction: column; gap: 15px;">
            <div style="background: #fdf2f2; padding: 15px; border-radius: 10px; border-left: 5px solid #e74c3c;">
                <b style="color: #e74c3c;">1. Approche Polaire (Points & Cercles)</b><br>
                Soit $z = r e^{i\\theta}$, alors $f(z) = r^2 e^{i(2\\theta)}$.
                <ul style="font-size: 13px; margin-top: 8px;">
                    <li><b>Module :</b> $|w| = r^2$. Un cercle de rayon $r$ devient un cercle de rayon $r^2$.</li>
                    <li><b>Argument :</b> $\\arg(w) = 2\\theta$. L'espace est "enroulé" deux fois sur lui-même.</li>
                </ul>
            </div>
            
            <div style="background: #e8f5e9; padding: 15px; border-radius: 10px; border-left: 5px solid #2e7d32; font-size: 13px;">
                <b style="color: #2e7d32;">Conséquence Dynamique :</b><br>
                Si $r < 1$, l'itération converge vers $0$.<br>
                Si $r > 1$, l'itération diverge vers $\\infty$.<br>
                La frontière critique est le cercle unité ($r=1$).
            </div>
        </div>

        <div style="flex: 1; background: #f0f7ff; padding: 15px; border-radius: 10px; border-left: 5px solid #2e86c1;">
            <b style="color: #2e86c1;">2. Approche Cartésienne (Droites)</b><br>
            En posant $z = x + iy$, on a $u = x^2 - y^2$ et $v = 2xy$.
            
            <p style="font-size: 13px; margin-top: 10px;"><b>Pour une droite verticale ($x = cste$) :</b></p>
            <p style="font-size: 14px; text-align: center; background: white; padding: 5px; border-radius: 5px;">
                $u = x^2 - \\left(\\frac{v}{2x}\\right)^2$
            </p>
            <p style="font-size: 13px;">On obtient une <b>parabole horizontale</b> s'ouvrant vers la gauche.</p>
            
            <p style="font-size: 13px; margin-top: 10px;"><b>Pour une droite horizontale ($y = cste$) :</b></p>
            <p style="font-size: 13px;">Le raisonnement symétrique mène à une parabole s'ouvrant vers la droite.</p>
        </div>
    </div>

    <div style="margin-top: 20px; padding: 10px; background: #2c3e50; color: white; border-radius: 8px; text-align: center; font-size: 15px;">
        <b>Bilan :</b> Ces déformations continues expliquent l'aspect "courbé" et "auto-similaire" des fractales de Julia.
    </div>
    """,
    prev_id="dynamique_plotly_z2_large"
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import io
import base64

# --- 1. GÉNÉRATION DU VISUEL (Cercle de période 2) ---
def get_period2_vis():
    t = np.linspace(0, 2*np.pi, 100)
    # Le bulbe de période 2 est un disque de centre -1 et de rayon 1/4
    c_re = -1 + 0.25 * np.cos(t)
    c_im = 0.25 * np.sin(t)
    
    plt.figure(figsize=(4, 4), facecolor='none')
    plt.plot(c_re, c_im, color='#f39c12', linewidth=3)
    plt.fill(c_re, c_im, color='#fef5e7', alpha=0.6)
    plt.axhline(0, color='black', lw=0.5)
    plt.axvline(-1, color='black', lw=0.5, ls='--')
    plt.title("Le Bulbe de Période 2", fontsize=10)
    plt.axis('equal')
    plt.grid(True, alpha=0.2)
    
    buf = io.BytesIO()
    plt.savefig(buf, format='png', bbox_inches='tight', transparent=True)
    plt.close()
    return base64.b64encode(buf.getvalue()).decode('utf-8')

img_p2 = get_period2_vis()

# --- 2. AFFICHAGE DE LA SLIDE ---
creer_slide_nav(
    section_id="analyse_periode_2",
    section_num="04",
    section_titre="Analyse Dynamique Profonde",
    sous_titre="Au-delà du Point Fixe : Le Cycle de Période 2",
    contenu=f"""
    <div style="display: flex; gap: 20px; align-items: start;">
        <div style="flex: 1.2;">
            <p>À gauche de la cardioïde se trouve un disque parfait. Il contient les paramètres $c$ où la suite $z_{{n+1}} = z_n^2 + c$ tend vers un <b>cycle limite de période 2</b>.</p>
            
            <div style="background: #fff3e0; padding: 12px; border-radius: 8px; border-left: 5px solid #f39c12; margin: 10px 0;">
                <b style="color: #e67e22;">Équation du cycle :</b><br>
                On cherche $z$ tel que $f_c(f_c(z)) = z$, soit $(z^2+c)^2+c = z$.<br>
                En factorisant par les points fixes d'ordre 1, il reste :<br>
                <p style="text-align:center; font-weight:bold;">$z^2 + z + (c + 1) = 0$</p>
            </div>
            
            <p style="font-size: 14px;">La condition de stabilité $|(f_c \circ f_c)'(z)| < 1$ se traduit par :</p>
            <p style="text-align: center; font-weight: bold; color: #d35400;">
                $|4z(z^2+c)| < 1 \implies |4(c+1)| < 1$
            </p>
        </div>
        
        <div style="flex: 0.8; text-align: center;">
            <img src="data:image/png;base64,{img_p2}" style="width: 100%; filter: drop-shadow(0px 4px 6px rgba(0,0,0,0.1));">
            <p style="font-size: 12px; color: #7f8c8d; margin-top: 10px;">Le disque de centre <b>-1</b> et de rayon <b>1/4</b></p>
        </div>
    </div>
    
    <div style="margin-top: 15px; padding: 10px; background: #f4f7f6; border-radius: 8px; font-size: 13px;">
        <b>Note théorique :</b> C'est le début de la "cascade de doublement de période" menant au chaos (Feigenbaum).
    </div>
    """,
    prev_id="synthese_algebrique_z2"
)
```

```python
creer_slide_nav(
    section_id="exploration_interactive_iframe",
    section_num="04",
    section_titre="Analyse Dynamique Profonde",
    sous_titre="Interface de Manipulation en Temps Réel",
    contenu="""
    <p>Cette interface affiche les bifurcations dynamiques de l'ensemble de Julia.</p>
    
    <div id="cadre_julia" style="width: 100%; border: 3px solid #2c3e50; border-radius: 12px; overflow: hidden; background-color: #000; position: relative;">
        <div style="background: #2c3e50; color: white; padding: 8px 15px; font-size: 13px; font-family: sans-serif; display: flex; justify-content: space-between;">
            <span>Moteur de Rendu Julia (Direct)</span>
            <span style="color: #27ae60;">● Connecté</span>
        </div>
        
        <canvas id="directJuliaCanvas" style="width:100%; height:450px; display:block;"></canvas>
        
        <div style="position:absolute; top:50px; left:15px; background:rgba(0,0,0,0.7); color:#00ff00; padding:8px; border-radius:4px; font-family:monospace; font-size:12px; border:1px solid #333;">
            c = <span id="valC"></span>
        </div>
    </div>

    <script>
    (function() {
        const canvas = document.getElementById('directJuliaCanvas');
        const ctx = canvas.getContext('2d');
        const valC = document.getElementById('valC');
        
        canvas.width = 600; 
        canvas.height = 400;

        let t = 0;
        function render() {
            const w = canvas.width;
            const h = canvas.height;
            const img = ctx.createImageData(w, h);
            t += 0.02;
            
            const cr = -0.7 + 0.2 * Math.cos(t);
            const ci = 0.27 + 0.2 * Math.sin(t * 0.8);
            if(valC) valC.innerText = cr.toFixed(3) + " + " + ci.toFixed(3) + "i";

            for (let x = 0; x < w; x++) {
                for (let y = 0; y < h; y++) {
                    let zx = 1.5 * (x - w/2)/(w/2);
                    let zy = (y - h/2)/(h/2);
                    let i = 50;
                    while(zx*zx + zy*zy < 4 && i > 0) {
                        let tmp = zx*zx - zy*zy + cr;
                        zy = 2*zx*zy + ci;
                        zx = tmp;
                        i--;
                    }
                    const p = (x + y * w) * 4;
                    img.data[p] = i * 5; 
                    img.data[p+1] = i * 3; 
                    img.data[p+2] = 200 - i * 3;
                    img.data[p+3] = 255;
                }
            }
            ctx.putImageData(img, 0, 0);
            requestAnimationFrame(render);
        }
        setTimeout(render, 500);
    })();
    </script>

    <div style="display: flex; gap: 15px; margin-top: 15px;">
        <div style="flex: 1; padding: 12px; background: #f4f7f9; border-radius: 8px; border-left: 5px solid #2c3e50; font-size: 13px;">
            <b>Analyse :</b> Transition entre ensembles connexes et poussières de Cantor.
        </div>
    </div>
    """,
    prev_id="constitution_mandelbrot"
)
```
